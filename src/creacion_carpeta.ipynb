```python
from pathlib import Path

# Carpeta raíz del proyecto
PROJECT_PATH = Path.cwd()

# Si estás ejecutando desde data/processed,
# subimos tres niveles hasta la raíz del proyecto
if PROJECT_PATH.name == "processed":
    PROJECT_PATH = PROJECT_PATH.parent.parent.parent

# Crear estructura
carpetas = [
    "src/data/raw",
    "src/data/processed",
    "src/notebooks",
    "src/utils",
    "presentacion",
    "memoria",
    "docs"
]

for carpeta in carpetas:
    (PROJECT_PATH / carpeta).mkdir(
        parents=True,
        exist_ok=True
    )

print("Estructura del proyecto creada:")
print(PROJECT_PATH)
```
