```python
# Revisar valores nulos, rangos, categorías, duplicados

import pandas as pd
from pathlib import Path

DATA_PATH = Path.cwd() / "data" / "raw"

# Cargar los cinco archivos y añadir el año
dataframes = []

for archivo in sorted(DATA_PATH.glob("*.csv")):
    df = pd.read_csv(
        archivo,
        sep=";",
        encoding="utf-8-sig",
        decimal=","
    )

    df["Año"] = int(archivo.stem[-5:-1])
    dataframes.append(df)

# Unificar los archivos
df_original = pd.concat(dataframes, ignore_index=True)

print("DIMENSIONES DEL DATASET")
print(df_original.shape)

print("\nVALORES NULOS POR COLUMNA")
display(df_original.isnull().sum().to_frame("Valores nulos"))

print("\nDUPLICADOS EXACTOS")
print(df_original.duplicated().sum())

print("\nESTADÍSTICAS DE LA VARIABLE NUMÉRICA")
display(df_original["Total"].describe())

print("\nVALORES FUERA DEL RANGO 0-100")
print(
    ((df_original["Total"] < 0) | (df_original["Total"] > 100)).sum()
)

print("\nNÚMERO DE CATEGORÍAS POR AÑO")
for columna in df_original.columns:
    print(f"\n{columna}:")
    print(df_original[columna].nunique())
```

```python
print("VALORES NULOS POR COLUMNA")
display(df_original.isnull().sum())

print("\nDUPLICADOS EXACTOS")
print(df_original.duplicated().sum())

print("\nREGISTROS POR AÑO")
display(
    df_original["Año"]
    .value_counts()
    .sort_index()
)

print("\nCATEGORÍAS DE TAMAÑO DE EMPRESA")
display(
    df_original["Tamaño de la empresa"]
    .value_counts()
)

print("\nNOMBRES DE LAS COLUMNAS")
print(df_original.columns.tolist())
```

```python
# Número de registros por año
print("REGISTROS POR AÑO")
display(
    df_original.groupby("Año").size().to_frame("Registros")
)

# Indicadores por año
print("\nNÚMERO DE INDICADORES POR AÑO")
display(
    df_original.groupby("Año")["Principales variables"]
    .nunique()
    .to_frame("Indicadores diferentes")
)

# Primeros indicadores únicos
print("\nLISTADO DE INDICADORES")
indicadores = sorted(
    df_original["Principales variables"].dropna().unique()
)

for indicador in indicadores:
    print(indicador)
```

```python
# Buscar indicadores que describen la adopción general de IA
mascara = df_original["Principales variables"].str.contains(
    "emplean tecnologías de Inteligencia Artificial",
    case=False,
    na=False
)

adopcion_general = df_original.loc[mascara].copy()

# Mostrar resultados por año, sector y tamaño
columnas = [
    "Año",
    "Principales variables",
    "Tamaño de la empresa",
    "Total"
]

# Añadimos la columna sectorial que corresponda a cada año
sector_anterior = "Agrupación de actividad (excepto CNAE 56, 64-66 y 95.1)"
sector_nuevo = "Agrupación de actividad (CNAE-25 sección C-O, excl. L y grupo 951)"

adopcion_general["Sector"] = adopcion_general[sector_anterior].fillna(
    adopcion_general[sector_nuevo]
)

display(
    adopcion_general[
        ["Año", "Sector", "Principales variables",
         "Tamaño de la empresa", "Total"]
    ].sort_values(["Año", "Sector", "Tamaño de la empresa"])
)
```

```python
# Crear una columna sectorial unificada
df_original["sector_raw"] = (
    df_original[
        "Agrupación de actividad (excepto CNAE 56, 64-66 y 95.1)"
    ]
    .combine_first(
        df_original[
            "Agrupación de actividad (CNAE-25 sección C-O, excl. L y grupo 951)"
        ]
    )
)

# Comprobar el resultado
print("Valores nulos en sector_raw:")
print(df_original["sector_raw"].isnull().sum())

print("\nNúmero de sectores por año:")
display(
    df_original.groupby("Año")["sector_raw"]
    .nunique()
    .to_frame("Número de sectores")
)

print("\nPrimeros sectores:")
display(
    df_original[["Año", "sector_raw"]]
    .drop_duplicates()
    .sort_values(["Año", "sector_raw"])
)
```

```python
print("Duplicados exactos:", df_original.duplicated().sum())

# Comprobar duplicados por combinación de variables
claves = [
    "Año",
    "sector_raw",
    "Principales variables",
    "Tamaño de la empresa"
]

print(
    "Duplicados por combinación de año, sector, indicador y tamaño:",
    df_original.duplicated(subset=claves).sum()
)
```

```python
# Crear dataset de trabajo
df = df_original[
    [
        "Año",
        "sector_raw",
        "Principales variables",
        "Tamaño de la empresa",
        "Total"
    ]
].copy()

# Renombrar columnas
df = df.rename(columns={
    "Año": "year",
    "sector_raw": "sector",
    "Principales variables": "indicator",
    "Tamaño de la empresa": "company_size",
    "Total": "value_pct"
})

# Ordenar registros
df = df.sort_values(
    ["year", "sector", "indicator", "company_size"]
).reset_index(drop=True)

# Comprobar estructura
print("Dimensiones:", df.shape)
print("\nColumnas:", df.columns.tolist())
display(df.head())

# Guardar dataset procesado
from pathlib import Path

OUTPUT_PATH = Path.cwd() / "data" / "processed"
OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

df.to_csv(
    OUTPUT_PATH / "dataset_ia_empresas_2022_2026.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nDataset guardado correctamente.")
```

```python
# Filtrar el indicador general de adopción de IA
mascara = df["indicator"].str.contains(
    "emplean tecnologías de Inteligencia Artificial",
    case=False,
    na=False
)

df_adopcion = df.loc[mascara].copy()

# Mostrar los registros de Total Empresas y Total
df_h1 = df_adopcion[
    (df_adopcion["sector"].str.contains(
        "Total Empresas", case=False, na=False
    )) &
    (df_adopcion["company_size"] == "Total")
].copy()

# Mostrar resultados
display(
    df_h1[["year", "indicator", "value_pct"]]
    .sort_values("year")
    .reset_index(drop=True)
)
```

```python
from IPython.display import Markdown, display

display(Markdown("""
## H1. Evolución de la adopción de inteligencia artificial

**Hipótesis:** La adopción de IA ha aumentado en las empresas españolas.

Se analiza el porcentaje de empresas que emplean tecnologías de IA entre 2022 y 2026, utilizando el total de empresas de cada año.
"""))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

df_h1 = df_h1.sort_values("year").copy()

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df_h1,
    x="year",
    y="value_pct",
    marker="o",
    linewidth=2.5,
    color="#2563EB"
)

for _, fila in df_h1.iterrows():
    plt.annotate(
        f"{fila['value_pct']:.2f}%",
        (fila["year"], fila["value_pct"]),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center"
    )

plt.title("Evolución de la adopción de IA en empresas españolas")
plt.xlabel("Año")
plt.ylabel("Empresas que utilizan IA (%)")
plt.xticks(df_h1["year"])
plt.ylim(0, 40)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
# Hipótesis H2: Las empresas grandes adoptan inteligencia artificial en mayor proporción que las pequeñas.

# Filtrar adopción general para el conjunto de empresas
df_h2 = df_adopcion[
    (df_adopcion["sector"] == "Total Empresas") &
    (df_adopcion["company_size"] != "Total")
].copy()

# Establecer el orden de los tamaños
orden_tamanos = [
    "De 10 a 49",
    "De 50 a 249",
    "De 250 y más"
]

df_h2["company_size"] = pd.Categorical(
    df_h2["company_size"],
    categories=orden_tamanos,
    ordered=True
)

# Tabla comparativa
tabla_h2 = df_h2.pivot(
    index="year",
    columns="company_size",
    values="value_pct"
)

display(tabla_h2.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(11, 6))

sns.barplot(
    data=df_h2,
    x="year",
    y="value_pct",
    hue="company_size",
    hue_order=[
        "De 10 a 49",
        "De 50 a 249",
        "De 250 y más"
    ],
    palette="Blues"
)

plt.title(
    "Adopción de inteligencia artificial según el tamaño de empresa",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Año")
plt.ylabel("Empresas que utilizan IA (%)")
plt.ylim(0, 85)
plt.legend(title="Tamaño de empresa")
plt.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()
```

```python
## H3. Adopción de inteligencia artificial por sector económico

# Seleccionar el año 2025 y el tamaño Total
df_h3 = df_adopcion[
    (df_adopcion["year"] == 2025) &
    (df_adopcion["company_size"] == "Total") &
    (df_adopcion["sector"] != "Total Empresas")
].copy()

# Ordenar sectores según su adopción de IA
df_h3 = df_h3.sort_values(
    "value_pct",
    ascending=False
)

print("Número de sectores:", len(df_h3))

display(
    df_h3[["sector", "value_pct"]]
    .reset_index(drop=True)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Crear una copia de los datos
df_h3_grafico = df_h3.copy()

# Nombres abreviados siguiendo el orden de los sectores por porcentaje
nombres_cortos = [
    "Información y comunicaciones",
    "Sector TIC",
    "Actividades profesionales y científicas",
    "Actividades inmobiliarias",
    "Total servicios",
    "Energía y agua",
    "Alojamiento",
    "Fabricación de productos informáticos",
    "Coquerías y refino de petróleo",
    "Servicios administrativos",
    "Comercio y reparación de vehículos",
    "Total industria",
    "Alimentación, textil y otras manufacturas",
    "Transporte y almacenamiento",
    "Metalurgia y productos metálicos",
    "Total construcción"
]

# Ordenar de mayor a menor
df_h3_grafico = df_h3_grafico.sort_values(
    "value_pct", ascending=False
).reset_index(drop=True)

df_h3_grafico["sector_corto"] = nombres_cortos

# Crear gráfico
fig, ax = plt.subplots(figsize=(12, 8))

sns.barplot(
    data=df_h3_grafico,
    x="value_pct",
    y="sector_corto",
    color="#2563EB",
    ax=ax
)

# Añadir porcentajes
for barra in ax.patches:
    valor = barra.get_width()

    ax.text(
        valor + 0.5,
        barra.get_y() + barra.get_height() / 2,
        f"{valor:.2f}%",
        va="center",
        fontsize=9
    )

ax.set_title(
    "Adopción de IA por sector económico en España (2025)",
    fontsize=14,
    fontweight="bold",
    pad=15
)

ax.set_xlabel("Empresas que utilizan IA (%)")
ax.set_ylabel("")
ax.set_xlim(0, 70)
ax.grid(axis="x", alpha=0.2)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
